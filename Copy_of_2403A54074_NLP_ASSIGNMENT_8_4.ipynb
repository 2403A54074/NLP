Unigram Count

```python
import collections

D1 = "Free fire players love rushing in Bermuda map"
D2 = "Professional free fire players practice every day"
D3 = "Rushing teams win more fights in free fire tournaments"
D4 = "I love playing free fire with my squad"

combined_words = f"{D1} {D2} {D3} {D4}"
words = combined_words.lower().split()

unigram_counts = collections.Counter(words)

print("Unigram Count:")
for word, count in unigram_counts.most_common():
    print(f"{word}: {count}")
```

Bigram Count

```python
import collections

combined_words = f"{D1} {D2} {D3} {D4}"
words = combined_words.lower().split()

bigrams = list(zip(words, words[1:]))

bigram_counts = collections.Counter(bigrams)

print("Bigram Count:")
for (w1, w2), count in bigram_counts.most_common():
    print(f"{w1} {w2}: {count}")
```

Trigram Count

```python
import collections

combined_words = f"{D1} {D2} {D3} {D4}"
words = combined_words.lower().split()

trigrams = list(zip(words, words[1:], words[2:]))

trigram_counts = collections.Counter(trigrams)

print("Trigram Count:")
for (w1, w2, w3), count in trigram_counts.most_common():
    print(f"{w1} {w2} {w3}: {count}")
```

Next Word Prediction using Bigram Count

```python
import collections

combined_words = f"{D1} {D2} {D3} {D4}"
words = combined_words.lower().split()

unigram_counts = collections.Counter(words)
bigrams = list(zip(words, words[1:]))
bigram_counts = collections.Counter(bigrams)

def bigram_probability(word_sequence, bigram_counts, unigram_counts):

    words_in_sequence = word_sequence.lower().split()

    if not words_in_sequence:
        return "Please provide a word sequence."

    last_word = words_in_sequence[-1]

    potential_next_words = {}

    for (w1, w2), count in bigram_counts.items():
        if w1 == last_word:
            potential_next_words[w2] = count

    if not potential_next_words:
        return f"No bigram found starting with '{last_word}'."

    last_word_unigram_count = unigram_counts.get(last_word, 0)

    predicted_word = None
    max_probability = -1

    for next_word, bigram_count in potential_next_words.items():
        probability = bigram_count / last_word_unigram_count
        print(f"Probability of '{next_word}' is {probability}")

        if probability > max_probability:
            max_probability = probability
            predicted_word = next_word

    return predicted_word
```

Test

```python
sequence1 = "free fire"
next_word1 = bigram_probability(sequence1, bigram_counts, unigram_counts)
print(f"Given sequence: '{sequence1}', predicted next word: '{next_word1}'")
```

Deployment of Bigram Model

```python
text = input("Enter some text: ")
next_word = bigram_probability(text, bigram_counts, unigram_counts)
print(f"Given sequence: '{text}', predicted next word: '{next_word}'")
```

Next Word Prediction using Trigram Count

```python
import collections

combined_words = f"{D1} {D2} {D3} {D4}".lower().split()

unigram_counts = collections.Counter(combined_words)
bigrams = list(zip(combined_words, combined_words[1:]))
bigram_counts = collections.Counter(bigrams)

trigrams = list(zip(combined_words, combined_words[1:], combined_words[2:]))
trigram_counts = collections.Counter(trigrams)

def predict_next_word_trigram(sequence, trigram_counts, bigram_counts):

    words_in_sequence = sequence.lower().split()

    if len(words_in_sequence) < 2:
        return "Sequence must contain at least two words for trigram prediction."

    last_two = tuple(words_in_sequence[-2:])
    potential_next_words = {}

    for (w1, w2, w3), count in trigram_counts.items():
        if (w1, w2) == last_two:
            potential_next_words[w3] = count

    if not potential_next_words:
        return f"No trigram found starting with '{' '.join(last_two)}'."

    last_two_count = bigram_counts.get(last_two, 0)

    predicted_word = None
    max_prob = -1

    for word, trigram_count in potential_next_words.items():
        probability = trigram_count / last_two_count
        print(f"Probability of '{word}' is {probability}")

        if probability > max_prob:
            max_prob = probability
            predicted_word = word

    return predicted_word
```

```python
sequence2 = "free fire players"
next_word2 = predict_next_word_trigram(sequence2, trigram_counts, bigram_counts)
print(f"Given sequence: '{sequence2}', predicted next word: '{next_word2}'")
```

Deployment of Trigram Model

```python
# Take user input
text = input("Enter at least two words: ")

# Predict next word using trigram
next_word = predict_next_word_trigram(text, trigram_counts, bigram_counts)

print(f"Given sequence: '{text}', predicted next word: '{next_word}'")
```

Next word prediction using Bigram with Laplace Smoothing

```python
V = len(unigram_counts)

def predict_next_word_bigram_laplace(sequence, bigram_counts, unigram_counts):

    words = sequence.lower().split()
    if not words:
        return "Please provide a word sequence."

    last_word = words[-1]
    candidates = {w2: count for (w1, w2), count in bigram_counts.items() if w1 == last_word}

    if not candidates:
        return f"No bigram found starting with '{last_word}'."

    last_word_count = unigram_counts.get(last_word, 0)

    predicted_word = None
    max_prob = -1

    for word, count in candidates.items():
        probability = (count + 1) / (last_word_count + V)
        print(f"Probability of '{word}' is {probability}")

        if probability > max_prob:
            max_prob = probability
            predicted_word = word

    return predicted_word
```
